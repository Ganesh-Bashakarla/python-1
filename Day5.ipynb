# slicing

```python
s='Python Programming'
print(s[3:10:])
print(s[3:10:2])
print(s[:5])
print(s[6:])
print(s[::-1])
```

# string operation

```python
s1='codegnan'
s2='students'
print(s1+s2)
print(s1*4)
print(s2*3)
print(s1<s2)
```

# string method 

```python
s1='stuDENts are good'
print(s1.lower())
print(s1.upper())
print(s1.capitalize())
print(s1.title())
print(s1.swapcase())
```

```python
s1='stuDENts are good'
print(s1.islower())
print(s1.isupper())
print(s1.isalnum())
print(s1.isalpha())
print(s1.isdigit())
print(s1.isspace())
print(' '.isspace())
```

```python
s1='stuDENts are good'
print(list(s1))
print(s1.split('#'))
print(s1.split())
print(' '.join(s1.split()))
```
